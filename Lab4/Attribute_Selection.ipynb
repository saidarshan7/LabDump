```python
import pandas as pd
import numpy as np

# Column names
columns = [
    "age","sex","cp","trestbps","chol","fbs","restecg",
    "thalach","exang","oldpeak","slope","ca","thal","num"
]

# Load dataset
df = pd.read_csv(
    "processed.switzerland.data",
    header=None,
    names=columns,
    na_values="?"
)

# Convert to numeric
df = df.apply(pd.to_numeric, errors="coerce")
df = df.dropna(how='all')


# Handle missing values
df.fillna(df.mean(), inplace=True)
df = df.astype(int)


print("First 5 rows:")
print(df.head())

print("\nDataset Shape:", df.shape)
```

```python
X = df.drop("num", axis=1)
y = df["num"]

print("\nIndependent variables shape:", X.shape)
print("Target variable shape:", y.shape)
```

```python
correlation_matrix = df.corr()

correlation_with_target = correlation_matrix["num"].drop("num")

feature_ranking = correlation_with_target.abs().sort_values(ascending=False)

print("\nFeature Ranking (Correlation with Target):")
print(feature_ranking)
```

```python
top_5_features = feature_ranking.head(5).index.tolist()

print("\nTop 5 Selected Features:")
print(top_5_features)
```

```python
X_top5 = df[top_5_features]

print("\nReduced Dataset Shape:", X_top5.shape)

print("\nFirst 5 rows of Reduced Dataset:")
print(X_top5.head())
```

```python
print(X_top5)
```

```python
df_reduced = X_top5.copy()
df_reduced["num"] = y.values

print(df_reduced.head())
print("Shape:", df_reduced.shape)
```
